## EDA 
### 1.1 Handling Missing Values

Features with a gap percentage of more than 50% were removed.

```python
null_sum = df.isnull().sum() / len(df)
columns_to_drop = null_sum[null_sum >= 0.5].index
modified_df = df.drop(columns=columns_to_drop)
modified_df.head()
```

### 1.2 Removing Low-Variance Features

Numerical features with CV below a threshold of some threshold and categorical features with a single category constitutes greater than 95% of observations were removed. ???

```python
std_num_features = modified_df.select_dtypes(include=['number']).std()
mean_num_features = modified_df.select_dtypes(include=['number']).mean()
coff_vars = std_num_features / mean_num_features
coff_vars[coff_vars < 0.1]
```

A correlation check with the target variable is performed for low-varianca features to prevent the removal of features that may have a predictive relationship

```python
low_cv_num_cols = modified_df[coff_vars[coff_vars < 0.1].index]
low_cv_num_cols
```
